### Importación de datos



````python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda.head()
````

#1. Análisis de facturación



````python
# Calcular el ingreso total de cada tienda
total_tienda1 = sum(tienda['Precio'])
total_tienda2 = sum(tienda2['Precio'])
total_tienda3 = sum(tienda3['Precio'])
total_tienda4 = sum(tienda4['Precio'])

print(f"Ingreso total de la tienda 1 {total_tienda1}")
print('######')
print(f"Ingreso total de la tienda 2 {total_tienda2}")
print('######')
print(f"Ingreso total de la tienda 3 {total_tienda3}")
print('######')
print(f"Ingreso total de la tienda 4 {total_tienda4}")

ingresos_totales = [total_tienda1, total_tienda2, total_tienda3, total_tienda4]
tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
````

````python
plt.figure(figsize=(8,5))
plt.bar(tiendas, ingresos_totales, color='skyblue', edgecolor='black')

plt.title('Ingresos totales por tienda')
plt.xlabel('Tienda')
plt.ylabel('Ingresos (en $)')
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
````

# 2. Ventas por categoría

````python
#calcular la cantidad de productos vendidos por categoría
productos_vendidos_tienda1 = tienda.groupby('Categoría del Producto')['Producto'].count()
productos_vendidos_tienda2 = tienda2.groupby('Categoría del Producto')['Producto'].count()
productos_vendidos_tienda3 = tienda3.groupby('Categoría del Producto')['Producto'].count()
productos_vendidos_tienda4 = tienda4.groupby('Categoría del Producto')['Producto'].count()

print("Total productos vendidos por la tienda 1")
print(productos_vendidos_tienda1)
print('######')
print("Total productos vendidos por la tienda 2")
print(productos_vendidos_tienda2)
print('######')
print("Total productos vendidos por la tienda 3")
print(productos_vendidos_tienda3)
print('######')
print("Total productos vendidos por la tienda 4")
print(productos_vendidos_tienda4)
print('######')
````

````python
plt.figure(figsize=(10,6))
productos_vendidos_tienda1.sort_values(ascending=True).plot(kind='barh', color='lightgreen', edgecolor='black')

plt.title('Cantidad de productos vendidos por categoría en la tienda 1')
plt.xlabel('Cantidad de ventas')
plt.ylabel('Categoría del Producto')
plt.tight_layout()
plt.show()
````

# 3. Calificación promedio de la tienda


````python
#calcular las calificaciones promedio de los clientes para cada tienda
promedio_tienda1 = tienda['Calificación'].mean()
promedio_tienda2 = tienda2['Calificación'].mean()
promedio_tienda3 = tienda3['Calificación'].mean()
promedio_tienda4 = tienda4['Calificación'].mean()

print(f"Calificación promedio de la tienda 1: {promedio_tienda1}")
print('######')
print(f"Calificación promedio de la tienda 2: {promedio_tienda2}")
print('######')
print(f"Calificación promedio de la tienda 3: {promedio_tienda3}")
print('######')
print(f"Calificación promedio de la tienda 4: {promedio_tienda4}")

tiendas = ["Tienda 1", "Tienda 2", "Tienda 3", "Tienda 4"]
promedios_totales = [promedio_tienda1, promedio_tienda2, promedio_tienda3, promedio_tienda4]
````

````python
plt.figure(figsize=(8,5))
plt.scatter(tiendas, promedios_totales, color='royalblue', s=100, edgecolor='black')

plt.title('Promedio de calificación por tienda')
plt.xlabel('Tiendas')
plt.ylabel('Promedio de calificación')
plt.ylim(3.5, 4.5)
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
````

# 4. Productos más y menos vendidos

````python
ventas_por_producto_tienda1 = tienda['Producto'].value_counts().reset_index()
ventas_por_producto_tienda1.columns = ['Producto', 'Cantidad_Ventas']
producto_mas_vendido_tienda1 = ventas_por_producto_tienda1.iloc[0]
producto_menos_vendido_tienda1 = ventas_por_producto_tienda1.iloc[-1]

ventas_por_producto_tienda2 = tienda2['Producto'].value_counts().reset_index()
ventas_por_producto_tienda2.columns = ['Producto', 'Cantidad_Ventas']
producto_mas_vendido_tienda2 = ventas_por_producto_tienda2.iloc[0]
producto_menos_vendido_tienda2 = ventas_por_producto_tienda2.iloc[-1]

ventas_por_producto_tienda3 = tienda3['Producto'].value_counts().reset_index()
ventas_por_producto_tienda3.columns = ['Producto', 'Cantidad_Ventas']
producto_mas_vendido_tienda3 = ventas_por_producto_tienda3.iloc[0]
producto_menos_vendido_tienda3 = ventas_por_producto_tienda3.iloc[-1]

ventas_por_producto_tienda4 = tienda4['Producto'].value_counts().reset_index()
ventas_por_producto_tienda4.columns = ['Producto', 'Cantidad_Ventas']
producto_mas_vendido_tienda4 = ventas_por_producto_tienda4.iloc[0]
producto_menos_vendido_tienda4 = ventas_por_producto_tienda4.iloc[-1]

print(f"Producto más vendido en la tienda 1: {producto_mas_vendido_tienda1}")
print(f"Producto menos vendido en la tienda 1: {producto_menos_vendido_tienda1}")
print("##########")
print(f"Producto más vendido en la tienda 2: {producto_mas_vendido_tienda2}")
print(f"Producto menos vendido en la tienda 2: {producto_menos_vendido_tienda2}")
print("##########")
print(f"Producto más vendido en la tienda 3: {producto_mas_vendido_tienda3}")
print(f"Producto menos vendido en la tienda 3: {producto_menos_vendido_tienda3}")
print("##########")
print(f"Producto más vendido en la tienda 4: {producto_mas_vendido_tienda4}")
print(f"Producto menos vendido en la tienda 4: {producto_menos_vendido_tienda4}")
print("##########")
````

# 5. Envío promedio por tienda

````python
#calcular el costo de envío promedio para cada tienda
costo_envio_promedio_tienda1 = tienda['Costo de envío'].mean()
costo_envio_promedio_tienda2 = tienda2['Costo de envío'].mean()
costo_envio_promedio_tienda3 = tienda3['Costo de envío'].mean()
costo_envio_promedio_tienda4 = tienda4['Costo de envío'].mean()

print(f"Costo envio promedio de la tienda 1: {costo_envio_promedio_tienda1}")
print('######')
print(f"Costo envio promedio de la tienda 2: {costo_envio_promedio_tienda2}")
print('######')
print(f"Costo envio promedio de la tienda 3: {costo_envio_promedio_tienda3}")
print('######')
print(f"Costo envio promedio de la tienda 4: {costo_envio_promedio_tienda4}")
````
